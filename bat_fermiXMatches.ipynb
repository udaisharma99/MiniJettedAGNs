```python
#gather all X-ray information available on CoreG and FR0 sources
import time
import logging
import numpy as np
from pathlib import Path
import astropy.units as u
from astropy.table import Table,  vstack 
from astropy.coordinates import Distance
from astropy.coordinates import SkyCoord, FK5
from astroquery.vizier import Vizier
from astroquery.xmatch import XMatch
```

```python
nagar_2005 = Vizier(columns=["**"], row_limit=-1).get_catalogs("J/A+A/435/521")
# - FR0 Catalog
fr0cat = Vizier(columns=["**"], row_limit=-1).get_catalogs("J/A+A/609/A1")
```

```python
bat157 = Table.read('BAT_157m.txt', format='ascii',delimiter='|')
# Rename 'BAT Name' to 'col2'
bat157.rename_column('col2', 'BAT Name')
bat157.rename_column('col3', 'RAJ2000')
bat157.rename_column('col4', 'DEJ2000')
```

```python
#4FGL-DR4 Catalog
fermidr4 = Table.read('4fgl-dr4.fit', format='fits')
fermidr4 = fermidr4['RAJ2000', 'DEJ2000','Source_Name']
```

```python
fermi_nagar_xmatches = XMatch.query(
    cat1=nagar_2005[0],
    cat2= fermidr4,
    colRA1="_RA",
    colDec1="_DE",
    colRA2="RAJ2000",
    colDec2="DEJ2000",
    max_distance=150 * u.arcsec,
)
fermi_fr0_xmatches = XMatch.query(
    cat1=fr0cat[0],
    cat2=fermidr4,
    colRA1="_RA",
    colDec1="_DE",
    colRA2="RAJ2000",
    colDec2="DEJ2000",
    max_distance=180 * u.arcsec,
)
```
