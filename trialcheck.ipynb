```python
import astropy
import numpy as np
from pathlib import Path
import astropy.units as u
from astropy.table import Table,  vstack 
from astropy.coordinates import Distance
from astropy.coordinates import SkyCoord, FK5
from astroquery.vizier import Vizier
```

```python
morx = Vizier(columns=["**"], row_limit=-1).get_catalogs("V/158/morxv2")
twosxps_swift = Vizier(columns=["**"], row_limit=-1).get_catalogs("IX/58/2sxps")
bat157 = Table.read('./data/catalogues/BAT_157m.txt', format='ascii',delimiter='|')
fourxmm = Table.read('./data/catalogues/4XMM_DR14cat_v1.0.csv', format='csv')
cxotwo = Vizier(columns=["**"], row_limit=-1).get_catalogs("IX/70/csc21mas")

```

```python
x_ray_catalogs = [morx[0], fourxmm, cxotwo[0], twosxps_swift[0], bat157]
```

```python
"""Search for X-ray counterparts in various catalogues -- MORX, 4XMM-DR14, CSC2.1, 2SXPS, BAT 157 Month Catalog.
Make sure x_ray_catalogs = [morx[0], fourxmm, cxotwo[0], twosxps_swift[0], bat157] """        
c = SkyCoord(ra=17.21*u.deg, dec=-0.655*u.deg, frame='icrs')
#Getting XMM, CXO and Swift counterparts from MORX
coords_morx = astropy.coordinates.SkyCoord(ra=x_ray_catalogs[0]['RAJ2000'],dec=x_ray_catalogs[0]['DEJ2000'],unit=(u.deg,u.deg))
crossmatch_morx = astropy.coordinates.match_coordinates_sky(c, coords_morx,nthneighbor=1)
morx_xmm = x_ray_catalogs[0][crossmatch_morx[0].item()]['XMM-ID']
morx_cxo = x_ray_catalogs[0][crossmatch_morx[0].item()]['CX-ID']
morx_swift = x_ray_catalogs[0][crossmatch_morx[0].item()]['Swift-ID']
morx_seperation = crossmatch_morx[1]
#4XMM-DR14 counterpart
coords_4xmm = astropy.coordinates.SkyCoord(ra=x_ray_catalogs[1]['ra'],dec=x_ray_catalogs[1]['dec'],unit=(u.deg,u.deg))
crossmatch_4xmm = astropy.coordinates.match_coordinates_sky(c, coords_4xmm,nthneighbor=1)
xmm_id = x_ray_catalogs[1][crossmatch_4xmm[0].item()]['iauname']
xmm_seperation = crossmatch_4xmm[1]
#CSC2.1 counterpart
coords_cxo = astropy.coordinates.SkyCoord(ra=x_ray_catalogs[2]['RAICRS'],dec=x_ray_catalogs[2]['DEICRS'],unit=(u.deg,u.deg))
crossmatch_cxo = astropy.coordinates.match_coordinates_sky(c, coords_cxo,nthneighbor=1)
cxo_id = x_ray_catalogs[2][crossmatch_cxo[0].item()]['2CXO']
cxo_seperation = crossmatch_cxo[1]
#2SXPS counterpart
coords_swift = astropy.coordinates.SkyCoord(ra=x_ray_catalogs[3]['RAJ2000'],dec=x_ray_catalogs[3]['DEJ2000'],unit=(u.deg,u.deg))
crossmatch_swift = astropy.coordinates.match_coordinates_sky(c, coords_swift,nthneighbor=1)
morx_swift = x_ray_catalogs[3][crossmatch_swift[0].item()]['IAUName']
morx_swift_seperation = crossmatch_swift[1]
#BAT 157 Month Catalog counterpart
coords_bat = astropy.coordinates.SkyCoord(ra=x_ray_catalogs[4]['col3'],dec=x_ray_catalogs[4]['col4'],unit=(u.deg,u.deg))
crossmatch_bat = astropy.coordinates.match_coordinates_sky(c, coords_bat,nthneighbor=1)
bat_id = x_ray_catalogs[4][crossmatch_bat[0].item()]['col2']
bat_seperation = crossmatch_bat[1]
```

```python
print(f"""
morx_xmm: {morx_xmm}
morx_cxo: {morx_cxo}
morx_swift: {morx_swift}
morx_source_seperation: {morx_seperation}/{morx_seperation.to(u.arcmin)}/{morx_seperation.to(u.arcsec)}
4xmm_id: {xmm_id}
4xmm_source_seperation: {xmm_seperation}/{xmm_seperation.to(u.arcmin)}/{xmm_seperation.to(u.arcsec)}
2cxo_id: {cxo_id}
2cxo_source_seperation: {cxo_seperation}/{cxo_seperation.to(u.arcmin)}/{cxo_seperation.to(u.arcsec)}
2sxps_swift_id: {morx_swift}
2sxps_swift_source_seperation: {morx_swift_seperation}/{morx_swift_seperation.to(u.arcmin)}/{morx_swift_seperation.to(u.arcsec)}
bat157month_id: {bat_id}
bat157month_source_seperation: {bat_seperation}/{bat_seperation.to(u.arcmin)}/{bat_seperation.to(u.arcsec)}
""")
```
